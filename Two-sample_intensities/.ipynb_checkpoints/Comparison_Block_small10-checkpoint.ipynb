```python
from tda import PD, PWGK, PL, PSSK

import tda
import numpy as np
import os
import random
import json
import math
from itertools import combinations
from scipy.stats import multivariate_normal
from joblib import Parallel, delayed
import itertools
from math import comb as math_comb
import pickle
```

```python
# compute MMDu
def n_mmd(mat_gram, unbias=True):
    n_total = mat_gram.shape[0]
    n = int(n_total / 2)
    mat_xx = mat_gram[0:n, 0:n]
    mat_yy = mat_gram[n:n_total, n:n_total]
    mat_xy = mat_gram[0:n, n:n_total]
    sum_xx = sum(sum(mat_xx))
    sum_yy = sum(sum(mat_yy))
    sum_xy = sum(sum(mat_xy))
    if unbias:
        sum_xx -= sum(np.diag(mat_xx))
        sum_yy -= sum(np.diag(mat_yy))
        sum_xy -= sum(np.diag(mat_xy))
        return (sum_xx + sum_yy - 2 * sum_xy) / (n - 1)
    else:
        return (sum_xx + sum_yy - 2 * sum_xy) / n


def hist_wchi(mat_gram, num_hist=int(1e+4)):
    n = len(mat_gram)

    # centered Gram matrix
    mat_center = np.empty((n, n))
    vec_gram = sum(mat_gram)
    val_total = sum(vec_gram)
    for i in range(n):
        for j in range(i + 1):
            mat_center[i, j] = (mat_gram[i, j]
                                - ((vec_gram[i] + vec_gram[j]) / n)
                                + (val_total / (n ** 2)))
            mat_center[j, i] = mat_center[i, j]

    # estimated eigenvalues
    vec_nu = np.sort(np.linalg.eigh(mat_center)[0])[::-1][0: - 1]
    vec_lambda = vec_nu / (n - 1)
    sum_lambda = sum(vec_lambda)

    # histogram of the null distribution (weighted chi square)
    vec_hist = np.empty(num_hist)
    for i in range(num_hist):
        vec_z = np.random.normal(0, np.sqrt(2), n - 1) ** 2
        vec_hist[i] = np.inner(vec_lambda, vec_z) - 2 * sum_lambda

    return np.sort(vec_hist)[::-1]


def extract_submat(mat_gram, num_m=None):
    n_total = mat_gram.shape[0]
    n = int(n_total / 2)
    if num_m is None:
        num_m = n - 1
    d = int(2 * num_m)
    mat = np.empty((d, d))
    idx_x = random.sample(range(0, n), num_m)
    idx_y = random.sample(range(n, n_total), num_m)
    idx_xy = idx_x + idx_y
    for i, a in enumerate(idx_xy):
        for j, b in enumerate(idx_xy):
            mat[i, j] = mat_gram[a, b]
    return mat


def two_sample_test(mat_gram, alpha=0.05, num_m=None, num_test=500):
    vec_wchi = hist_wchi(mat_gram)                    # null distribution of psi-hat
    vec_p_value = np.empty(num_test)
    for temp_test in range(num_test):                 # for l=1,...,N
        mat_reduced = extract_submat(mat_gram, num_m)  # resample m samples
        value_mmd = n_mmd(mat_reduced)                 # compute mMMDu
        vec_temp = np.where(vec_wchi > value_mmd)[0]   # how many psi-hat's are greater than mMMDu?
        vec_p_value[temp_test] = len(vec_temp) / len(vec_wchi)
    return vec_p_value, len(np.where(vec_p_value < alpha)[0]) / num_test


```

```python
###BLOCK###

def epanechnikov_kernel_2d(diffs):
    # Epanechnikov kernel in 2D
    norm_sq = np.sum(diffs**2, axis=1)
    inside = norm_sq <= 1
    return 3 / 4 * (1 - norm_sq[inside])

def compute_g_fast(mu, nu, kernel='gaussian', sigma=np.array([[1, 0], [0, 1]])):
    idx_mu = np.repeat(np.arange(mu.shape[0]), nu.shape[0])
    idx_nu = np.tile(np.arange(nu.shape[0]), mu.shape[0])
    diffs = mu[idx_mu] - nu[idx_nu]
    
    if kernel == 'gaussian':
        densities = multivariate_normal.pdf(diffs, mean=[0, 0], cov=sigma)
    elif kernel == 'Epan':
        densities = np.zeros(diffs.shape[0])
        densities[:len(epanechnikov_kernel_2d(diffs))] = epanechnikov_kernel_2d(diffs)
    else:
        raise ValueError("Unknown kernel type")
    
    return np.sum(densities)

def make_U_fast(diagram_list_1, diagram_list_2, sigma, cores=4, kernel='gaussian'):
    N1 = len(diagram_list_1)
    N2 = len(diagram_list_2)

    comb1 = list(itertools.combinations(range(N1), 2))
    comb2 = list(itertools.combinations(range(N2), 2))
    grid3 = [(i, j) for i in range(N1) for j in range(N2)]

    def g_fast_pair(i, j, list1, list2):
        return compute_g_fast(list1[i], list1[j], kernel=kernel, sigma=sigma)

    psi_1 = sum(Parallel(n_jobs=cores)(
        delayed(g_fast_pair)(i, j, diagram_list_1, diagram_list_1) for i, j in comb1)) / (N1 * (N1 - 1) / 2)

    psi_2 = sum(Parallel(n_jobs=cores)(
        delayed(g_fast_pair)(i, j, diagram_list_2, diagram_list_2) for i, j in comb2)) / (N2 * (N2 - 1) / 2)

    psi_12 = sum(Parallel(n_jobs=cores)(
        delayed(compute_g_fast)(diagram_list_1[i], diagram_list_2[j], kernel=kernel, sigma=sigma)
        for i, j in grid3)) / (N1 * N2)

    return psi_1 + psi_2 - 2 * psi_12

def smart_split(lst, l1):
    n = len(lst)
    q, r = divmod(n, l1)
    sizes = [l1] * (q - r) + [l1 + 1] * r
    result = []
    index = 0
    for size in sizes:
        result.append(lst[index: index + size])
        index += size
    return result
    
def BlockU(list1, list2, lambda_, t, d, kernel, sigma):
    n1 = len(list1)
    n2 = len(list2)
    l1 = int(n1 ** lambda_)
    l2 = int(n2 ** lambda_)
    
    splitted_list1 = smart_split(list1, l1)
    splitted_list2 = smart_split(list2, l2)
    
    m = min(len(splitted_list1), len(splitted_list2))
    I = np.zeros(m)
    length_list = np.zeros(m)
    
    for mm in range(m):
        sub1 = splitted_list1[mm]
        sub2 = splitted_list2[mm]
        lj1 = len(sub1)
        lj2 = len(sub2)

        g_values_1 = np.zeros((lj1, lj1))
        for i in range(lj1):
            for j in range(i+1,lj1):
                val = compute_g_fast(sub1[i], sub1[j], kernel=kernel, sigma=sigma)
                g_values_1[i, j]= val
                g_values_1[j, i]= val 
        g_values_2 = np.zeros((lj2, lj2))
        for i in range(lj2):
            for j in range(i+1,lj2):
                val = compute_g_fast(sub2[i], sub2[j], kernel=kernel, sigma=sigma)
                g_values_2[i, j]=val
                g_values_2[j, i]=val

        g_values_12 = np.zeros((lj1, lj2))
        for i in range(lj1):
            for j in range(lj2):
                g_values_12[i, j] = compute_g_fast(sub1[i], sub2[j], kernel=kernel, sigma=sigma)

        case1 = list(combinations(range(lj1), 2))
        case2 = list(combinations(range(lj2), 2))
        total = 0
        for i1, i2 in case1:
            for j1, j2 in case2:
                total += (
                    g_values_1[i1, i2] + g_values_2[j1, j2]
                    - 0.5 * (
                        g_values_12[i1, j1] + g_values_12[i1, j2] +
                        g_values_12[i2, j1] + g_values_12[i2, j2]
                    )
                )
        #print((total / (comb(lj1, 2) * comb(lj2, 2))) * ((lj1 + lj2) ** ((d + 1) / 2)))
        I[mm] = (total / (math_comb(lj1, 2) * math_comb(lj2, 2))) * ((lj1 + lj2) ** ((d + 1) / 2))
        length_list[mm] = lj1 + lj2

    theta_hat = make_U_fast(list1, list2, sigma=sigma, cores=4, kernel=kernel)
    I_tilde = I - np.array([x ** ((d + 1) / 2) * theta_hat for x in length_list])

    Cmt = list(combinations(range(m), t))
    St = np.mean([np.prod([length_list[idx] ** ((d + 1) / 2) for idx in combo]) for combo in Cmt])
    Tmt = np.mean([np.prod([I_tilde[idx] for idx in combo]) for combo in Cmt]) + St * theta_hat ** t

    return Tmt, m

def compute_Histogram_BLOCKU(list1, list2, B=1000, Z_samples=10000, sigma=np.array([[1, 0],  [0, 1]]), random_state=42):
    np.random.seed(random_state)

    m = len(list1)
    n = len(list2)

    # g(mu, nu) 계산 결과 저장
    g_values = np.zeros((m, n))
    for i in range(m):
        for j in range(n):
            g_values[i, j] = compute_g_fast(list1[i], list2[j], sigma=sigma)

    g_flat = g_values.flatten()
    V1 = np.var(g_flat, ddof=1)

    # C_hat 계산
    cov_vals = np.zeros(B)
    g_mean = np.mean(g_flat)

    for b in range(B): #We only use list1 for estimating parameter. 
        i = np.random.choice(m)
        ip = np.random.choice([x for x in range(m) if x != i])
        j = np.random.choice([x for x in range(m) if x not in [i,ip]])

        val1 = compute_g_fast(list1[i], list1[j], sigma=sigma)
        val2 = compute_g_fast(list1[ip], list1[j], sigma=sigma)

        cov_vals[b] = (val1 - g_mean) * (val2 - g_mean)

    C_hat = np.mean(cov_vals)
    V_total = V1 - 2 * C_hat

    r = m / (m + n)
    s = 1 - r

    v_square = (1 / r**2) * 2 * V_total + (1 / s**2) * 2 * V_total + (1 / (r * s)) * 16 * (1/4) * V_total

    # Hermite H_2(Z) = Z^2 - 1
    
    Z = np.random.normal(size=Z_samples)
    H2_Z = (Z**2 - 1) * v_square

    return H2_Z

def two_sample_test_BLOCKU(list1,list2, alpha=0.05, num_m=None, num_test=500):
    vec_wchi = compute_Histogram_BLOCKU(list1,list2,sigma=np.array([[1/2, 0],  [0, 1/2]]))   # null distribution of psi-hat
    vec_p_value = np.empty(num_test)
    for temp_test in range(num_test):                 # for l=1,...,N
        sublist1= random.sample(list1,num_m//2)  # resample m//2 samples
        sublist2= random.sample(list2,num_m//2)  # resample m//2 samples
        value_BLOCK = BlockU(sublist1, sublist2, lambda_=1/2, t=2, d=1, kernel='gaussian',sigma=np.array([[1/2, 0],  [0, 1/2]])) # compute BlockU
        vec_temp = np.where(vec_wchi > value_BLOCK[0]*value_BLOCK[1])[0]   # how many psi-hat's are greater than value_BLOCK?
        vec_p_value[temp_test] = len(vec_temp) / len(vec_wchi)
    return vec_p_value, len(np.where(vec_p_value < alpha)[0]) / num_test

from joblib import Parallel, delayed

def compute_single_I(sub1, sub2, d, kernel, sigma):
    lj1, lj2 = len(sub1), len(sub2)
    idx1 = np.array(list(itertools.combinations(range(lj1), 2)))
    idx2 = np.array(list(itertools.combinations(range(lj2), 2)))
    
    total = 0.0
    for i1, i2 in idx1:
        for j1, j2 in idx2:
            total += h_kernel(sub1[i1], sub1[i2], sub2[j1], sub2[j2], kernel, sigma)
    
    norm_factor = math.comb(lj1, 2) * math.comb(lj2, 2)
    value = total / norm_factor * (lj1 + lj2) ** ((d + 1) / 2)
    length = lj1 + lj2
    return value, length

```

```python
# simulation parameters
npc=100
nset=500
nsig=4
```

```python
# import PD - json, each pds are saved as unnamed array

with open("twopd.json") as f:
    twopd = json.load(f)

twopdlist=[]
for ii in range(len(twopd)):
    twopdsublist = []
    for jj in range(len(twopd[0])):
        twopddat = np.array(twopd[ii][jj])
        twopdmat = np.transpose( np.resize(twopddat, (3,int(len(twopddat)/3)) ) )
        twopddim1 = twopdmat[twopdmat[:,0]==1,1:]
        twopdsublist.append(twopddim1)
    twopdlist.append(twopdsublist)
```

```python
comb = np.array(list(combinations(range(nset), 2)))
random.seed(2)
ind=random.sample(range(len(comb)),nset)
```

```python
# functions for PWGK
func_kernel = tda.function_kernel("Gaussian", sigma=(0.5) )
```

## constant weight

```python
# no weight
func_weight = tda.function_weight("none")
perz_cons=np.zeros((nsig,nset))
for ii in range(nsig):
    print(ii)
    for jj in range(nset):
        # import dimension one PDs
        combpdlist = twopdlist[ii][npc*(comb[ind[jj]][0]):npc*(comb[ind[jj]][0]+1)]
        combpdlist.extend(twopdlist[ii][npc*(comb[ind[jj]][1]):npc*(comb[ind[jj]][1]+1)])
        # compute gram matrix
        pwgk = PWGK(combpdlist, func_kernel, func_weight, sigma=(0.5),approx=True)
        mat_gaussian_pwgk = pwgk.gram_matrix()
        # define gram matrix
        name_rkhs = ["Linear", "Gaussian"][1]
        mat_gram_pwgk = tda.matrix_gram(mat_gaussian_pwgk, name_rkhs)[0]
        num_reject = two_sample_test(mat_gram_pwgk, alpha=0.05, num_m=10, num_test=500)[1]
        perz_cons[ii,jj]=num_reject
```

```python
1-np.sum(perz_cons<0.05,axis=1)/nset # constant
```
