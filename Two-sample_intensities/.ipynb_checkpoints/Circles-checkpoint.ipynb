```python
from tda import PD, PWGK, PL, PSSK

import tda
import numpy as np
import os
import random
import json
import pandas as pd
from itertools import combinations
import pickle
import pickle
from scipy.spatial.distance import pdist
def savepkl(obj, path):
    """객체를 피클 파일로 저장"""
    with open(path, 'wb') as f:
        pickle.dump(obj, f)

def loadpkl(path):
    """피클 파일에서 객체 불러오기"""
    with open(path, 'rb') as f:
        return pickle.load(f)

import time
```

```python
# compute MMDu
def n_mmd(mat_gram, unbias=True):
    n_total = mat_gram.shape[0]
    n = int(n_total / 2)
    mat_xx = mat_gram[0:n, 0:n]
    mat_yy = mat_gram[n:n_total, n:n_total]
    mat_xy = mat_gram[0:n, n:n_total]
    sum_xx = sum(sum(mat_xx))
    sum_yy = sum(sum(mat_yy))
    sum_xy = sum(sum(mat_xy))
    if unbias:
        sum_xx -= sum(np.diag(mat_xx))
        sum_yy -= sum(np.diag(mat_yy))
        sum_xy -= sum(np.diag(mat_xy))
        return (sum_xx + sum_yy - 2 * sum_xy) / (n - 1)
    else:
        return (sum_xx + sum_yy - 2 * sum_xy) / n


def hist_wchi(mat_gram, num_hist=int(1e+4)):
    n = len(mat_gram)

    # centered Gram matrix
    mat_center = np.empty((n, n))
    vec_gram = sum(mat_gram)
    val_total = sum(vec_gram)
    for i in range(n):
        for j in range(i + 1):
            mat_center[i, j] = (mat_gram[i, j]
                                - ((vec_gram[i] + vec_gram[j]) / n)
                                + (val_total / (n ** 2)))
            mat_center[j, i] = mat_center[i, j]

    # estimated eigenvalues
    vec_nu = np.sort(np.linalg.eigh(mat_center)[0])[::-1][0: - 1]
    vec_lambda = vec_nu / (n - 1)
    sum_lambda = sum(vec_lambda)

    # histogram of the null distribution (weighted chi square)
    vec_hist = np.empty(num_hist)
    for i in range(num_hist):
        vec_z = np.random.normal(0, np.sqrt(2), n - 1) ** 2
        vec_hist[i] = np.inner(vec_lambda, vec_z) - 2 * sum_lambda

    return np.sort(vec_hist)[::-1]


def extract_submat(mat_gram, num_m=None):
    n_total = mat_gram.shape[0]
    n = int(n_total / 2)
    if num_m is None:
        num_m = n - 1
    d = int(2 * num_m)
    mat = np.empty((d, d))
    idx_x = random.sample(range(0, n), num_m)
    idx_y = random.sample(range(n, n_total), num_m)
    idx_xy = idx_x + idx_y
    for i, a in enumerate(idx_xy):
        for j, b in enumerate(idx_xy):
            mat[i, j] = mat_gram[a, b]
    return mat


def two_sample_test(mat_gram, alpha=0.05, num_m=None, num_test=500):
    vec_wchi = hist_wchi(mat_gram)                    # null distribution of psi-hat
    vec_p_value = np.empty(num_test)
    for temp_test in range(num_test):                 # for l=1,...,N
        mat_reduced = extract_submat(mat_gram, num_m)  # resample m samples
        value_mmd = n_mmd(mat_reduced)                 # compute mMMDu
        vec_temp = np.where(vec_wchi > value_mmd)[0]   # how many psi-hat's are greater than mMMDu?
        vec_p_value[temp_test] = len(vec_temp) / len(vec_wchi)
    return vec_p_value, len(np.where(vec_p_value < alpha)[0]) / num_test

def compute_sigma_median(D_list):
    """
    Compute median heuristic bandwidth for a list of persistence diagrams.

    Parameters:
        D_list (list of ndarray): Each element is an (n_i x 2) numpy array for a diagram.

    Returns:
        float: Median of within-diagram pairwise distances.
    """
    sigma_list = []
    for D in D_list:
        if D.shape[0] < 2:
            continue  # Cannot compute distance
        dist_vals = pdist(D, metric='euclidean')
        sigma_list.append(np.median(dist_vals))

    if len(sigma_list) == 0:
        return np.nan
    else:
        return np.median(sigma_list)
```

```python
# simulation parameters
npc=10
nset=500
nsig=4
sig=[0.05,0.10,0.15,0.20]
```

```python
# import PD - json, each pds are saved as unnamed array

with open("twopd.json") as f:
    twopd = json.load(f)

twopdlist=[]
for ii in range(len(twopd)):
    twopdsublist = []
    for jj in range(len(twopd[0])):
        twopddat = np.array(twopd[ii][jj])
        twopdmat = np.transpose( np.resize(twopddat, (3,int(len(twopddat)/3)) ) )
        twopddim1 = twopdmat[twopdmat[:,0]==1,1:]
        twopdsublist.append(twopddim1)
    twopdlist.append(twopdsublist)
```

```python
comb = np.array(list(combinations(range(nset), 2)))
random.seed(2)
ind=random.sample(range(len(comb)),nset)
```

```python
# functions for PWGK
func_kernel = tda.function_kernel("Gaussian", sigma=(0.5) )
```

## constant weight

```python
# no weight
func_weight = tda.function_weight("none")
perz_cons=np.zeros((nsig,nset))
for ii in range(nsig):
    print(ii)
    for jj in range(nset):
        # import dimension one PDs
        combpdlist = twopdlist[ii][npc*(comb[ind[jj]][0]):npc*(comb[ind[jj]][0]+1)]
        combpdlist.extend(twopdlist[ii][npc*(comb[ind[jj]][1]):npc*(comb[ind[jj]][1]+1)])
        # compute gram matrix
        rff_sigma = compute_sigma_median(combpdlist)
        func_kernel = tda.function_kernel("Gaussian", sigma=rff_sigma)
        pwgk = PWGK(combpdlist, func_kernel, func_weight, sigma=rff_sigma,approx=True)
            
        mat_gaussian_pwgk = pwgk.gram_matrix()
        # define gram matrix
        name_rkhs = ["Linear", "Gaussian"][1]
        mat_gram_pwgk = tda.matrix_gram(mat_gaussian_pwgk, name_rkhs)[0]
        num_reject = two_sample_test(mat_gram_pwgk, alpha=0.05, num_m=10, num_test=1000)[1]
        perz_cons[ii,jj]=num_reject
```

```python
1-np.sum(perz_cons<0.05,axis=1)/nset # constant
```

# Sesitivity to noise

```python
# simulation parameters
npc=300
nset=100
sig=[0.03,0.04,0.05,0.06,0.07]
nsig=len(sig)
```

```python
# import PD - json, each pds are saved as unnamed array

with open("npc=300_nset=100onepd_moresigma.json") as f:
    onepd = json.load(f)
with open("npc=300_nset=100twopd_moresigma.json") as f:
    twopd = json.load(f)

# convert json (saved as unnamed array) to python list
onepdlist=[]
for ii in range(len(onepd)):
    onepdsublist = []
    for jj in range(len(onepd[0])):
        onepddat = np.array(onepd[ii][jj])
        onepdmat = np.transpose( np.resize(onepddat, (3,int(len(onepddat)/3)) ) )
        onepddim1 = onepdmat[onepdmat[:,0]==1,1:]
        onepdsublist.append(onepddim1)
    onepdlist.append(onepdsublist)
    
twopdlist=[]
for ii in range(len(twopd)):
    twopdsublist = []
    for jj in range(len(twopd[0])):
        twopddat = np.array(twopd[ii][jj])
        twopdmat = np.transpose( np.resize(twopddat, (3,int(len(twopddat)/3)) ) )
        twopddim1 = twopdmat[twopdmat[:,0]==1,1:]
        twopdsublist.append(twopddim1)
    twopdlist.append(twopdsublist)
```

```python
comb = np.array(list(combinations(range(nset), 2)))
random.seed(42)
ind=random.sample(range(len(comb)),nset)
# linear
func_weight = tda.function_weight("linear", lin_el=1)
perz_sensitivity=pd.DataFrame(columns=['Noise1','Noise2','Power'])
for kk in range(nsig):
    empirical_pvalue=[]
    for jj in range(nset):
        # import dimension one PDs
        combpdlist = twopdlist[2][npc*(comb[ind[jj]][0]):npc*(comb[ind[jj]][0]+1)]
        combpdlist.extend(twopdlist[kk][npc*(comb[ind[jj]][1]):npc*(comb[ind[jj]][1]+1)])
        # compute gram matrix
        rff_sigma = compute_sigma_median(combpdlist)
        func_kernel = tda.function_kernel("Gaussian", sigma=rff_sigma)
        pwgk = PWGK(combpdlist, func_kernel, func_weight, sigma=rff_sigma,approx=True)
        mat_gaussian_pwgk = pwgk.gram_matrix()
        # define gram matrix
        name_rkhs = ["Linear", "Gaussian"][1]
        mat_gram_pwgk = tda.matrix_gram(mat_gaussian_pwgk, name_rkhs)[0]
        vec_wchi = hist_wchi(mat_gram_pwgk)
        value_nmmd = n_mmd(mat_gram_pwgk)
        vec_temp = np.where(vec_wchi > value_nmmd)[0]
        empirical_pvalue=np.append(empirical_pvalue,(len(vec_temp) / len(vec_wchi)))
    new_row = {'Noise1':np.round(sig[2],3) , 'Noise2': np.round(sig[kk],3), 'Power': np.sum(empirical_pvalue<0.05)/nset}
    perz_sensitivity.loc[len(perz_sensitivity)] = new_row
```

# Power and Time(s)

```python
# import PD - json, each pds are saved as unnamed array

with open("onepd_different_samplesize.json") as f:
    onepd = json.load(f)
with open("twopd_different_samplesize.json") as f:
    twopd = json.load(f)

# convert json (saved as unnamed array) to python list
onepdlist=[]
for ii in range(len(onepd)):
    onepdsublist = []
    for jj in range(len(onepd[ii])):
        onepddat = np.array(onepd[ii][jj])
        onepdmat = np.transpose( np.resize(onepddat, (3,int(len(onepddat)/3)) ) )
        onepddim1 = onepdmat[onepdmat[:,0]==1,1:]
        onepdsublist.append(onepddim1)
    onepdlist.append(onepdsublist)

twopdlist=[]
for ii in range(len(twopd)):
    twopdsublist = []
    for jj in range(len(twopd[ii])):
        twopddat = np.array(twopd[ii][jj])
        twopdmat = np.transpose( np.resize(twopddat, (3,int(len(twopddat)/3)) ) )
        twopddim1 = twopdmat[twopdmat[:,0]==1,1:]
        twopdsublist.append(twopddim1)
    twopdlist.append(twopdsublist)
```

```python
# simulation parameters
npc=[200,400,600,800,1000]
nset=100


comb = np.array(list(combinations(range(nset), 2)))
random.seed(42)
ind=random.sample(range(len(comb)),nset)
```

```python
# linear
random.seed(42)
func_weight = tda.function_weight("linear", lin_el=1)
perz_power_time=pd.DataFrame(columns=['Sample_size','Power','Time(s)'])
for kk in range(len(npc)):
    empirical_pvalue=[]
    empirical_time=[]
    for jj in range(nset):
        start=time.time()
        combpdlist = onepdlist[kk][npc[kk]*(comb[ind[jj]][0]):npc[kk]*(comb[ind[jj]][0]+1)]
        combpdlist.extend(twopdlist[kk][npc[kk]*(comb[ind[jj]][1]):npc[kk]*(comb[ind[jj]][1]+1)])
        
        rff_sigma = compute_sigma_median(combpdlist)
        func_kernel = tda.function_kernel("Gaussian", sigma=rff_sigma)
        pwgk = PWGK(combpdlist, func_kernel, func_weight, sigma=rff_sigma,approx=True)
        mat_gaussian_pwgk = pwgk.gram_matrix()
        
        name_rkhs = ["Linear", "Gaussian"][1]
        mat_gram_pwgk = tda.matrix_gram(mat_gaussian_pwgk, name_rkhs)[0]
        vec_wchi = hist_wchi(mat_gram_pwgk[:npc,:npc]) #Approximation of null distribution
        value_nmmd = n_mmd(mat_gram_pwgk)
        vec_temp = np.where(vec_wchi > value_nmmd)[0] 
        empirical_pvalue=np.append(empirical_pvalue,(len(vec_temp) / len(vec_wchi)))
        elapsed = time.time()-start
        empirical_time.append(elapsed)
    new_row = {'Sample_size':npc[kk] ,  'Power': np.sum(empirical_pvalue<0.05)/nset,'Time(s)':np.sum(empirical_time)/nset}
    perz_power_time.loc[len(perz_power_time)] = new_row
```
